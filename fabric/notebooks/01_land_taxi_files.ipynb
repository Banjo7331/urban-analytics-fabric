```python
workspace = "urban-analytics"
window_start = "2024-01-01"
window_end = "2024-06-30"
```

```python
from datetime import date

from urban_analytics.platform import Platform, Window

platform = Platform.fabric(workspace, Window(date.fromisoformat(window_start), date.fromisoformat(window_end)))
```

```python
import os

from urban_analytics.ingest import land_taxi_files

LAKEHOUSE_FILES = "/lakehouse/default/Files"


def open_sink(relative_path):
    target = f"{LAKEHOUSE_FILES}/{relative_path}"
    os.makedirs(os.path.dirname(target), exist_ok=True)
    return open(target, "wb")


landed = land_taxi_files(platform, open_sink)
display(spark.createDataFrame([(item.relative_path, item.bytes) for item in landed], ["file", "bytes"]))
```
